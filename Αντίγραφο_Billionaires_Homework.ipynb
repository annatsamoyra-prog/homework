## 7) Ποιες εταιρίες έχουν τους περισσότερους δισεκατομμυριούχους? Κάντε ένα γράφημα με τους top 5. ένα horizontal bar graph.

>
> **Συμβουλή:** Μπορείτε να δοκιμάσετε το `.head()` αντί για όλο το `df`
>

```python
top_companies = df['company'].value_counts().head(5)
top_companies
```

```python
import matplotlib.pyplot as plt
top_companies.plot(kind='barh')
plt.xlabel("Αριθμός billionaires")
plt.ylabel("Εταιρία")
plt.title("Top 5 εταιρίες με τους περισσότερους billionaires")
plt.show()
```

## 12) Φτιάξτε ένα scatterplot που να σχετίζει το net worth σε σχέση με την ηλικία τους.

```python
df.plot(kind='scatter', x='age', y='networthusbillion')
plt.xlabel("Ηλικία")
plt.ylabel("Πλούτος (δισ. $)")
plt.title("Ηλικία vs Πλούτος")
plt.show()
```

## 13) Φτιάξτε ένα γράφημα με τους top 10 πλουσιότερους billionaires

```python
top10 = df.sort_values(by='networthusbillion', ascending=False).head(10)

top10.plot(kind='barh', x='name', y='networthusbillion')
plt.xlabel("Πλούτος (δισ. $)")
plt.ylabel("Όνομα")
plt.title("Top 10 Πλουσιότεροι Billionaires")
plt.gca().invert_yaxis()
plt.show()
```
